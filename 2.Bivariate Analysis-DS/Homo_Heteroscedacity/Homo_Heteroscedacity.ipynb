```python
### 1.Import necessary libraries
```

```python
import pandas as pd #importing pandas libraries as pd where pandas have all the functions required for the dataset
```

2. DATA preprocessing

```python
dataset=pd.read_csv("Salary_Data.csv") #assigning the provided dataset to "dataset" variable in read form
dataset #checking whether we can fetch the datasets
```

3.Input and Output Split

```python
independent=dataset[["YearsExperience"]] #assigning the input coloumn from the dataset to 'independent' variable
independent 
```

```python
dependent=dataset[["Salary"]]#assigning the output coloumn from the dataset to 'dependent' variable
dependent
```

4.Train and Test Split

```python
from sklearn.model_selection import train_test_split #from sklearn.model_selection filename, we are importing train_test_split function
x_train,x_test,y_train,y_test= train_test_split(independent, dependent, test_size=0.30, random_state=0) #splitting the train and test datasets and 
#assigning them to separate variable names|| test size is the percentage of test data need to get split 
x_train,x_test,y_train,y_test
```

5.Model Creation

```python
from sklearn.linear_model import LinearRegression #from sklearn.linear_model filename, we are importing LinearRegression function
regressor=LinearRegression() #Assigning Linearregression function to regressor variable
regressor.fit(x_train,y_train) #fit is used for model creation
```

#constant values should be found

```python
weight=regressor.coef_ 
weight
```

```python
bias=regressor.intercept_
bias
```

6. Test set

```python
#From the created model, we need to find the predicted value
y_pred=regressor.predict(x_test)
```

7.Evaluation metrics

```python
residuals = y_test - y_pred # to actual output - predicted output
```

```python
import matplotlib.pyplot as plt
plt.scatter(y_pred, residuals)
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residuals vs. Predicted Values')
plt.show()
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
r_score
```

8.Save the best model

```python
import pickle
filename="finalized_model_linear.sav"
```

9.Load the saved model--->get inputs--->predict--->Call to action

```python
pickle.dump(regressor,open(filename,'wb'))
```

```python
loaded_model=pickle.load(open("finalized_model_linear.sav",'rb'))
```
